```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

print("1. A carregar o dataset original...")
df = pd.read_csv('diabetes.csv')

# Separar características (X) e a variável alvo (y - Outcome)
X = df.drop('Outcome', axis=1)
y = df['Outcome']

# 2. Divisão dos dados em Treino (80%) e Teste (20%) ANTES de qualquer tratamento
print("2. A dividir os dados em Treino e Teste (80/20)...")
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 3. Tratamento de zeros anómalos (imputação baseada EXPLICITAMENTE no conjunto de treino)
colunas_com_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Substituir 0 por NaN em ambos os conjuntos
X_treino[colunas_com_zeros] = X_treino[colunas_com_zeros].replace(0, np.nan)
X_teste[colunas_com_zeros] = X_teste[colunas_com_zeros].replace(0, np.nan)

# Calcular a média estritamente utilizando as linhas de TREINO
medias_treino = X_treino[colunas_com_zeros].mean()

# Preencher os valores nulos tanto no Treino quanto no Teste usando as médias do Treino
X_treino[colunas_com_zeros] = X_treino[colunas_com_zeros].fillna(medias_treino)
X_teste[colunas_com_zeros] = X_teste[colunas_com_zeros].fillna(medias_treino)
print("3. Zeros anómalos tratados com sucesso (usando as médias do treino).")

# 4. Normalização Min-Max (0 a 1) com ajuste estrito no treino
scaler = MinMaxScaler()
X_treino_normalizado = scaler.fit_transform(X_treino) # Fit e Transform no treino
X_teste_normalizado = scaler.transform(X_teste)      # Apenas Transform no teste

# Converter de volta para DataFrames do Pandas
X_treino_df = pd.DataFrame(X_treino_normalizado, columns=X.columns)
X_teste_df = pd.DataFrame(X_teste_normalizado, columns=X.columns)

# Juntar a coluna 'Outcome' de volta aos dataframes finais
df_treino_final = X_treino_df.copy()
df_treino_final['Outcome'] = y_treino.values

df_teste_final = X_teste_df.copy()
df_teste_final['Outcome'] = y_teste.values

# 5. Exportar as tabelas tratadas para ficheiros CSV para o Git
df_treino_final.to_csv('diabetes_treino_processado.csv', index=False)
df_teste_final.to_csv('diabetes_teste_processado.csv', index=False)

print("\nPré-processamento concluído com sucesso!")
print("Ficheiros gerados e prontos para o repositório:")
print(" - diabetes_treino_processado.csv")
print(" - diabetes_teste_processado.csv")
```
