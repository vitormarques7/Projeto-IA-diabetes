```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler


# 1. CARREGAR OS DADOS

df = pd.read_csv('diabetes.csv')

print("--- PRIMEIRAS LINHAS DOS DADOS ORIGINAIS ---")
display(df.head())

# 2. VERIFICAÇÃO DE DADOS TEXTUAIS/CATEGÓRICOS
# Ao imprimir os tipos, provamos que a base já é 100% numérica (int64 e float64).

print("\n--- TIPOS DE DADOS NAS COLUNAS ---")
print(df.dtypes)

# 3. TRATAMENTO DE DADOS AUSENTES (ZEROS ANÓMALOS)
# variáveis como IMC e Espessura da Pele não podem ser zero num paciente, tratando isso como falha na recolha.
colunas_com_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Substituir o valor numérico 0 por NaN (Not a Number) para o Pandas detetar a falha
df[colunas_com_zeros] = df[colunas_com_zeros].replace(0, np.nan)

# Preencher os valores NaN com a média exata da respetiva coluna, conforme o artigo
df[colunas_com_zeros] = df[colunas_com_zeros].fillna(df[colunas_com_zeros].mean())

print("\n--- CONTAGEM DE VALORES NULOS APÓS TRATAMENTO ---")
print(df.isnull().sum())

# 4. NORMALIZAÇÃO MIN-MAX (AJUSTE DE ESCALA)
# Separar a base entre Características (X) e a Resposta/Diagnóstico (y)
X = df.drop('Outcome', axis=1)
y = df['Outcome']

# Inicializar e aplicar o algoritmo de normalização Min-Max
scaler = MinMaxScaler()
X_normalizado = scaler.fit_transform(X)

# Voltar a transformar a matriz (numpy array) num DataFrame do Pandas para visualizar melhor
X_normalizado_df = pd.DataFrame(X_normalizado, columns=X.columns)

print("\n--- RESUMO ESTATÍSTICO APÓS NORMALIZAÇÃO MIN-MAX ---")
# O describe() vai provar ao teu professor que o valor 'min' é 0 e o 'max' é 1 em todas as colunas
display(X_normalizado_df.describe())

print("\n--- PRIMEIRAS LINHAS DOS DADOS TOTALMENTE PRÉ-PROCESSADOS ---")
display(X_normalizado_df.head())
from google.colab import files

# dataframe para CSV
X_normalizado_df.to_csv('diabetes_pre_processado.csv', index=False)

files.download('diabetes_pre_processado.csv')
```
