```python
import pandas as pd
import numpy as np

# 1. Carregar os dados originais
df = pd.read_csv('diabetes.csv')

# 2. Tratamento de zeros anômalos (substituição pela média)
# Apenas estas colunas clínicas não podem ter valor zero
colunas_com_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Substituir o 0 por NaN e depois preencher com a média
df[colunas_com_zeros] = df[colunas_com_zeros].replace(0, np.nan)
df[colunas_com_zeros] = df[colunas_com_zeros].fillna(df[colunas_com_zeros].mean())

# 3. Exportar os dados tratados para um novo arquivo
df.to_csv('diabetes_sem_escala.csv', index=False)

print("Tratamento concluído!")
print("Arquivo 'diabetes_sem_escala.csv' gerado com sucesso.")

from google.colab import files
files.download('diabetes_sem_escala.csv')

```
